```python
'''Modelo de Gradiente'''

for sample in samples:
    for s in sets:
        version='best'#8000//(sample//batch)-1
        #print('version: ',version)
        model='model_{sample}_set_{s}'.format(sample=sample,s=s)
        artifact_name='iie-sc/EGNN_gradient/'+model+':{version}'.format(version=version)
        print(artifact_name)
        #print('model: ',model)
        dir = "artifacts/"+ artifact_name.split('/')[1]  
        artifact = run.use_artifact(artifact_name , type='model')
        artifact_dir = artifact.download()    
        # Ruta original y nueva del archivo descargado
        original_path = os.path.join(artifact_dir, 'model.ckpt')
        #dir="artifacts/EGNN_MNF/"
        new_path = os.path.join(dir, model+'.ckpt')
        print('original: ',original_path)
        print('new: ',new_path)
        # Renombrar el archivo
        os.rename(original_path, new_path)
    
```
